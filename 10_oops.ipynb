### 🔹 1. Introduction to OOP (Classes and Objects)
Object-Oriented Programming (OOP) is a paradigm that uses **Classes** and **Objects** to structure programs by grouping related data and behaviors into individual entities.

* **Class:** A blueprint, template, or prototype from which objects are created.
* **Object:** An instance of a class. It has state (attributes) and behavior (methods).
* **Constructor (`__init__`):** A special initialization method called automatically when an object is instantiated.
* **`self`:** Represents the specific instance of the class. It connects the methods to the object's attributes.

```python
class Student:
    # Constructor
    def __init__(self, name, age, grade):
        self.name = name   # Instance Attribute
        self.age = age     # Instance Attribute
        self.grade = grade # Instance Attribute
        
    # Instance Method
    def display_info(self):
        return f"Student: {self.name} | Age: {self.age} | Grade: {self.grade}"

# Instantiating objects
s1 = Student("Kamran", 22, "A")
s2 = Student("Ayesha", 21, "A+")

print(s1.display_info())
print(s2.display_info())
```

### 🔹 2. Instance vs Class vs Static Variables & Methods

#### 🔹 1. Variables
* **Instance Variables:** Defined inside constructors, unique to each instance.
* **Class Variables:** Defined directly inside the class but outside any methods, shared by all instances.

```python
class Employee:
    # Class Variable (shared by all employees)
    company_name = "Tech Corp"
    
    def __init__(self, name, salary):
        # Instance Variables (unique to each employee)
        self.name = name
        self.salary = salary

emp1 = Employee("Alice", 80000)
emp2 = Employee("Bob", 90000)

print(emp1.name, "-", emp1.company_name)
print(emp2.name, "-", emp2.company_name)
```

#### 🔹 2. Methods
* **Instance Methods:** Normal class methods. Take `self` as the first argument. Can access and modify instance attributes.
* **Class Methods:** Decorated with `@classmethod`. Take `cls` as the first argument. Can access/modify class-level variables or act as factory methods.
* **Static Methods:** Decorated with `@staticmethod`. Do not take `self` or `cls`. Behave like normal functions but belong inside the class namespace for logical grouping.

```python
class DateUtil:
    # Class variable
    century = "21st"
    
    def __init__(self, day, month, year):
        self.day = day
        self.month = month
        self.year = year
        
    # Class Method
    @classmethod
    def from_string(cls, date_str):
        # Acts as an alternative constructor/factory
        day, month, year = map(int, date_str.split("-"))
        return cls(day, month, year)
        
    # Static Method
    @staticmethod
    def is_valid_year(year):
        return 1900 <= year <= 2100

# Using Class Method factory
date1 = DateUtil.from_string("11-06-2026")
print(f"Instantiated date: {date1.day}/{date1.month}/{date1.year}")

# Using Static Method
print("Is 2026 valid?", DateUtil.is_valid_year(2026))
```

### 🔹 3. Inheritance
Inheritance allows a child (sub) class to acquire the properties and methods of a parent (super) class, promoting code reuse.

#### Types of Inheritance:
1. **Single:** One child inherits from one parent.
2. **Multiple:** One child inherits from multiple parents.
3. **Multilevel:** A child inherits from a class, which itself inherits from another class.
4. **Hierarchical:** Multiple children inherit from a single parent.
5. **Hybrid:** A combination of two or more inheritance types.

#### 🧠 Method Resolution Order (MRO) & `super()`
* **`super()`:** Calls parent methods or constructors.
* **MRO (Method Resolution Order):** The order in which Python searches for a method in a class hierarchy. Python uses the **C3 Linearization** algorithm. You can check this using `ClassName.mro()`.

```python
class Person:
    def __init__(self, name):
        self.name = name
    def greet(self):
        return f"Hello, I am {self.name}."

class Speaker:
    def speak(self):
        return "Speaking in public..."

# Multiple Inheritance
class Lecturer(Person, Speaker):
    def __init__(self, name, subject):
        # Call Person constructor
        super().__init__(name)
        self.subject = subject
        
    def lecture(self):
        return f"Lecturing on {self.subject}."

prof = Lecturer("Dr. Khan", "Machine Learning")
print(prof.greet())
print(prof.speak())
print(prof.lecture())
print("MRO of Lecturer:", Lecturer.mro())
```

### 🔹 4. Polymorphism
Polymorphism means "many forms". It allows methods in different classes to have the same name but different behaviors.

#### 🔹 1. Method Overriding
A subclass redefines a method that exists in its superclass.

```python
class Animal:
    def make_sound(self):
        return "Some generic sound"

class Dog(Animal):
    def make_sound(self):
        return "Woof Woof!"

class Cat(Animal):
    def make_sound(self):
        return "Meow!"

animals = [Dog(), Cat(), Animal()]
for animal in animals:
    print(animal.__class__.__name__, ":", animal.make_sound())
```

#### 🔹 2. Operator Overloading (Dunder/Magic Methods)
* Python allows you to redefine how operators (`+`, `-`, `*`, `==`, etc.) work when applied to custom objects.
* This is done by implementing special methods prefixed and suffixed with double underscores (dunder methods).

| Operator | Dunder Method | Purpose |
| :---: | :--- | :--- |
| `+` | `__add__(self, other)` | Add two custom objects |
| `*` | `__mul__(self, other)` | Multiply two custom objects |
| `str()` / `print()` | `__str__(self)` | User-friendly string representation |
| `repr()` | `__repr__(self)` | Developer/debug string representation |
| `len()` | `__len__(self)` | Return length of custom object |

```python
class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y
        
    # Overload the '+' operator
    def __add__(self, other):
        return Vector(self.x + other.x, self.y + other.y)
        
    # Overload str()
    def __str__(self):
        return f"({self.x}, {self.y})"
        
    # Overload repr()
    def __repr__(self):
        return f"Vector({self.x}, {self.y})"

v1 = Vector(2, 4)
v2 = Vector(5, -1)
v3 = v1 + v2 # Internally calls: v1.__add__(v2)

print("Vector sum:", v3)
print("Vector repr:", repr(v3))
```

### 🔹 5. Encapsulation
Encapsulation wraps data (attributes) and methods into a single unit and restricts direct access to some of the object's components to prevent unintended modification.

#### 🔹 Access Modifiers in Python
1. **Public:** Access is open from anywhere. Example: `self.name`
2. **Protected (Convention only):** Indicated by single underscore `_`. Should not be accessed outside the class hierarchy. Example: `self._salary`
3. **Private (Enforced):** Indicated by double underscores `__`. Triggers **Name Mangling** to make direct external access difficult. Example: `self.__pin`

#### 🔹 Property Decorators (`@property`)
Instead of Java-like `get_balance()` and `set_balance()` methods, Python uses property decorators to create clean getter/setter syntax.

```python
class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        self._account_type = "Savings"
        self.__balance = balance # Private attribute
        
    # Getter using @property
    @property
    def balance(self):
        return self.__balance
        
    # Setter with validation
    @balance.setter
    def balance(self, amount):
        if amount < 0:
            raise ValueError("Balance cannot be negative!")
        self.__balance = amount

acc = BankAccount("Kamran", 5000)
print("Account Owner:", acc.owner)
print("Balance:", acc.balance) # Call getter

acc.balance = 6000 # Call setter
print("New Balance:", acc.balance)

# Try accessing private variable directly -> Raises AttributeError
try:
    print(acc.__balance)
except AttributeError as e:
    print("Cannot access private directly:", e)

# Name Mangling work-around (not recommended in practice!)
print("Accessing mangled name:", acc._BankAccount__balance)
```

### 🔹 6. Abstraction
Abstraction hides implementation details and only shows essential features to the user.

* Python implements abstraction using the **`abc`** (Abstract Base Classes) module.
* Classes inheriting from `ABC` with at least one `@abstractmethod` are **abstract classes** and cannot be instantiated.
* Subclasses **must** override all abstract methods to be instantiable.

```python
from abc import ABC, abstractmethod

class Vehicle(ABC):
    @abstractmethod
    def start_engine(self):
        pass

class Car(Vehicle):
    def start_engine(self):
        return "Car engine started: Vroom Vroom!"

class Truck(Vehicle):
    def start_engine(self):
        return "Truck engine started: Rumblleee..."

# vehicle = Vehicle() # Raises TypeError

c = Car()
t = Truck()
print(c.start_engine())
print(t.start_engine())
```
