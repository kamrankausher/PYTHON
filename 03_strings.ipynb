#### STRINGS ARE IMMUTABLE

```python
s = "hello"
s[0] = "H"   # ❌ ERROR
```

Why?

Strings cannot be changed in place.

Instead:

```python
s = "H" + s[1:]
print(s)
```

**Why immutability matters :**

* Safe for concurrency
* Predictable behavior
* Enables caching
* Important for LLM prompt safety

```python
#Q24 Write a program to perform basic run-length compression on a string (e.g., 'aaabbc' -> 'a3b2c1').
def compress(s):
    if not s: return ''
    result = []
    count = 1
    for i in range(1, len(s)):
        if s[i] == s[i-1]:
            count += 1
        else:
            result.append(s[i-1] + str(count))
            count = 1
    result.append(s[-1] + str(count))
    return ''.join(result)
print(compress('aaabbc'))
```

```python
#Q27 Write a program to find all occurrences of a substring inside a string and output their indices.
s = 'abracadabra'
sub = 'ab'
idx = 0
while True:
    idx = s.find(sub, idx)
    if idx == -1: break
    print('Found at:', idx)
    idx += 1
```

```python
#Q29 Write a program to implement basic template rendering replacement (e.g. replacing `{name}` with a value manually).
template = 'Hello {name}, your code is {code}.'
context = {'name': 'Amit', 'code': '1234'}
rendered = template
for k, v in context.items():
    rendered = rendered.replace('{' + k + '}', v)
print(rendered)
```
