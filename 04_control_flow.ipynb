# Control Flow (if, elif, else)

#### What is Control Flow? (First principles)

Control flow determines:

* Which code runs
* Under what condition
* In what order

Without control flow:

* Programs would always run line-by-line
* No decisions
* No intelligence

**👉 AI systems are nothing but complex decision systems**

*** 

### if Statement
🔹 Syntax

if condition:
    code_block

```python
loss = 0.05

if loss < 0.1:
    print("Model is learning")
```

##### 🧠 Important rule

* Condition must evaluate to boolean
* Indentation defines scope (very important in Python)

**Examples**

```python
n = "Sum"
m = "Sum"
if n==m:
    print("Both are same")
```

```python
loss= 0.2
accuracy = 0.9
if loss < 0.4 and accuracy > 0.7:
    print("Model is performing well")
```

***

### if-else

Used when exactly one path must run.

```python
if accuracy > 0.9:
    print("Good model")
else:
    print("Needs improvement")
```

```python
#Q Accept two numbers and print the greatest between them.

num_1 = int(input("Enter a first number"))
num_2 = int(input("Enter a second number"))
if num_1 > num_2:
    print(f"{num_1} is Greater than {num_2}")
elif num_2 > num_1:
    print(f"{num_2} is Greater than {num_1}")
else:
    print("Both numbers are same")
```

### if-elif-else (Multiple decisions)

Used when you have multiple mutually exclusive conditions

```python
score = 87
if score >= 90:
    grade = "A"
elif score >= 75:
    grade = "B"
elif score >= 60:
    grade = "C"
else:
    grade = "Fail"
print(grade)
```

##### 🧠 Execution rule

* Checked top to bottom
* First True block executes
* Rest are skipped


 

***

### Nested Conditions (Conditions inside conditions)

```python
model_trained =True
if model_trained:
    if accuracy > 0.9:
        print("Deploy model")
    else:
        print("Tune hyperparameters")
```

***

### Truthy & Falsy Values

**🔻 Falsy values in Python**

These evaluate to False:
* False
* None
* 0
* 0.0
* "" (empty string)
* [] (empty list)
* {} (empty dict)
* set()
  
 ***Everything else is Truthy.***

**🧠 Example**

```python
data = []

if data:
    print("Data exists")
else:
    print("No data")
```

Used heavily in:

* Data validation
* ML pipelines
* API checks

***

### Real Decision Logic in ML Systems

**🔹 Example 1: Early Stopping**

if loss < threshold and epoch > min_epoch:
    stop_training = True

**🔹 Example 2: Data Validation**

if not data:
    raise ValueError("Dataset is empty")

**🔹 Example 3: Model Selection**

if accuracy > best_accuracy:
    best_model = current_model

**🔹 Example 4: Inference Logic**

if confidence > 0.8:
    return "Accepted"
else:
    return "Rejected"
