# Regular Expressions (Regex)

***

### 🔹 1. Introduction to Regex & Python Raw Strings
A **Regular Expression (Regex)** is a sequence of characters that forms a search pattern. It is used to check if a string contains a specified search pattern, extract matches, or clean text.

#### 🔹 Python Raw Strings (`r"..."`)
In regular expressions, backslashes `\` are used frequently (e.g. `\d`, `\w`). However, Python strings use the backslash as an escape character (e.g. `\n` represents a newline).
* **Rule:** Always use **raw strings** (prefixed with `r`) when writing regex patterns to prevent Python from escaping backslashes.

```python
# Normal string vs Raw string
normal_str = "Hello\nWorld"
raw_str = r"Hello\nWorld"

print("Normal String:")
print(normal_str)
print("-" * 20)
print("Raw String:")
print(raw_str)
```

### 🔹 2. The re Module API
Python has a built-in package called `re` to work with regular expressions. Here are the core methods:

| Method | Description |
| :--- | :--- |
| **`re.search(pattern, string)`** | Searches for the first match anywhere in the string. Returns a Match object. |
| **`re.match(pattern, string)`** | Checks for a match **only at the beginning** of the string. Returns a Match object. |
| **`re.findall(pattern, string)`** | Finds all occurrences of the pattern. Returns a list of strings. |
| **`re.finditer(pattern, string)`** | Finds all occurrences and returns an iterator of Match objects (memory efficient). |
| **`re.sub(pattern, repl, string)`** | Replaces matches with a replacement string. |
| **`re.split(pattern, string)`** | Splits the string by occurrences of the pattern. |

```python
import re

text = "The quick brown fox jumps over 2 lazy dogs. The fox is fast."

# 1. re.search()
match_search = re.search(r"fox", text)
if match_search:
    print(f"Found '{match_search.group()}' at index: {match_search.start()} to {match_search.end()}")

# 2. re.match() - Searches only at start
match_match = re.match(r"fox", text)
print("Match at start?", match_match) # None, because text starts with 'The'

# 3. re.findall() - Returns list of matches
print("Findall fox:", re.findall(r"fox", text))

# 4. re.sub() - Replace
cleaned_text = re.sub(r"fox", "cat", text)
print("Substituted:", cleaned_text)

# 5. re.split() - Split by digits
print("Split by digits:", re.split(r"\d", text))
```

### 🔹 3. Regex Syntax & Metacharacters
Metacharacters are characters with special meanings:

#### 🔹 1. Character Classes:
* **`.`**: Any character except newline.
* **`\d`**: Any decimal digit (equivalent to `[0-9]`).
* **`\D`**: Any non-digit.
* **`\w`**: Any alphanumeric character (letters, numbers, underscore `_`).
* **`\W`**: Any non-alphanumeric character.
* **`\s`**: Any whitespace character (spaces, tabs, newlines).
* **`\S`**: Any non-whitespace character.

#### 🔹 2. Quantifiers:
* **`*`**: Zero or more occurrences.
* **`+`**: One or more occurrences.
* **`?`**: Zero or one occurrence (optional).
* **`{n}`**: Exactly `n` occurrences.
* **`{min, max}`**: Between `min` and `max` occurrences.

#### 🔹 3. Anchors & Boundaries:
* **`^`**: Matches start of string.
* **`$`**: Matches end of string.
* **`\b`**: Word boundary.

```python
# Matching a Phone Number: 3 digits, dash, 3 digits, dash, 4 digits (e.g. 123-456-7890)
phone_text = "Call me at 555-867-5309 or office line 111-222-3333."
phone_pattern = r"\d{3}-\d{3}-\d{4}"

phones = re.findall(phone_pattern, phone_text)
print("Found Phone Numbers:", phones)
```

```python
# Matching custom character sets
# Matches words starting with capital letters A-M
text = "Alice and Bob went to Delhi with Zakir."
pattern = r"\b[A-M]\w*"
print("Matched names starting with A-M:", re.findall(pattern, text))
```

### 🔹 4. Groups & Capturing
Parenthesis `()` are used to group sub-patterns together and extract specific fields from a matching string.

* **`group(0)`**: Returns the entire match.
* **`group(1)`**: Returns the first captured group.
* **`groupdict()`**: Returns a dictionary of Named Groups (`(?P<name>pattern)`).

```python
# Extracting username and domain from emails
email_text = "Send reports to admin@techcorp.com or support@sales.org"
email_pattern = r"(\w+)@(\w+\.\w+)"

# Using finditer to iterate over match objects
for match in re.finditer(email_pattern, email_text):
    print(f"Full Email: {match.group(0)}")
    print(f"  Username: {match.group(1)}")
    print(f"  Domain:   {match.group(2)}")
```

```python
# Using Named Groups (highly readable)
log_entry = "2026-06-11 10:15:30 [ERROR] Database connection failed."
log_pattern = r"(?P<date>\d{4}-\d{2}-\d{2}) (?P<time>\d{2}:\d{2}:\d{2}) \[(?P<level>\w+)\] (?P<message>.*)"

match = re.search(log_pattern, log_entry)
if match:
    data = match.groupdict()
    print("Extracted Log Data:")
    print("Date:", data["date"])
    print("Level:", data["level"])
    print("Message:", data["message"])
```

### 🔹 5. Greedy vs Lazy Matching
* By default, quantifiers (`*`, `+`, `{}`) are **greedy** — they match as much text as possible.
* Adding a **`?`** after the quantifier makes it **lazy** (non-greedy) — it matches as little text as possible.

```python
html_content = "<div>Title</div><div>Content</div>"

# 1. Greedy matching (matches from first <div> to last </div>)
greedy_pattern = r"<div>.*</div>"
print("Greedy:", re.findall(greedy_pattern, html_content))

# 2. Lazy matching (matches individual <div> tags separately)
lazy_pattern = r"<div>.*?</div>"
print("Lazy:", re.findall(lazy_pattern, html_content))
```

### 🔹 6. Real-world Data Cleaning for NLP / Data Prep
In NLP pipelines, we must clean raw text (like tweets, user feedback) by removing URLs, hashtags, punctuation, and extra whitespace before tokens are converted into embeddings.

```python
def clean_tweet(tweet):
    # 1. Remove URLs (http/https)
    tweet = re.sub(r"https?://\S+", "", tweet)
    # 2. Remove Twitter handles (@username)
    tweet = re.sub(r"@\w+", "", tweet)
    # 3. Remove hashtags but keep the text (e.g. #ML -> ML)
    tweet = re.sub(r"#(\w+)", r"\1", tweet)
    # 4. Remove punctuation and special characters
    tweet = re.sub(r"[^\w\s]", "", tweet)
    # 5. Remove duplicate/extra whitespace
    tweet = re.sub(r"\s+", " ", tweet).strip()
    return tweet

raw_tweet = "Awesome tutorial on #ML and #NLP! Check it out at https://example.com/blog @kamran_dev !!!"
print("Cleaned Tweet:", clean_tweet(raw_tweet))
```
