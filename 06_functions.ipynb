### 📦 Variable-Length Arguments (*args & **kwargs)
Sometimes we don't know beforehand how many arguments a user might pass. Python provides `*args` and `**kwargs` for this.

#### 🔹 1. *args (Non-keyword / Positional arguments)
* The `*` unpacking operator gathers positional arguments into a **tuple**.
* Useful when writing utility functions like summing numbers or joining text strings.

```python
def sum_all(*args):
    print(type(args), args)
    total = 0
    for num in args:
        total += num
    return total

print("Sum:", sum_all(1, 2, 3, 4, 5))
```

#### 🔹 2. **kwargs (Keyword arguments)
* The `**` unpacking operator gathers named/keyword arguments into a **dictionary**.
* Widely used in ML configurations and API calls to pass hyperparameters dynamically.

```python
def print_hyperparameters(**kwargs):
    print(type(kwargs), kwargs)
    for model_param, value in kwargs.items():
        print(f"Hyperparameter - {model_param}: {value}")

print_hyperparameters(learning_rate=0.01, epochs=100, optimizer="adam")
```

#### 🔹 3. Mixing Parameter Types
* **Rule of order:** Positional parameters first → `*args` → Default parameters → `**kwargs`.
* Breaking this order results in a `SyntaxError`.

```python
def train_model(model_name, *args, epochs=10, **kwargs):
    print(f"Training Model: {model_name}")
    print(f"Positional args: {args}")
    print(f"Epochs: {epochs}")
    print(f"Additional settings (kwargs): {kwargs}")

train_model("RandomForest", "dataset_v1", "verbose_mode", epochs=50, max_depth=10, min_samples_split=2)
```

### 📝 Docstrings & Function Annotations (Type Hinting)
Good code is self-documenting. In industry settings (and large ML codebases), docstrings and type hints are mandatory.

#### 🔹 1. Docstrings (`__doc__`)
* Triple-quoted strings placed immediately after the function header.
* Explains the function's purpose, parameters, and return value.
* Accessible via `help(function_name)` or `function_name.__doc__`.

```python
def calculate_bmi(weight: float, height: float) -> float:
    """
    Calculate the Body Mass Index (BMI).
    
    Parameters:
    weight (float): Weight in kilograms.
    height (float): Height in meters.
    
    Returns:
    float: Calculated Body Mass Index.
    """
    return weight / (height ** 2)

# Accessing documentation
print(calculate_bmi.__doc__)
```

#### 🔹 2. Type Hinting / Annotations
* Introduced in Python 3.5 to suggest types for parameters (`param: type`) and return value (`-> return_type`).
* **Note:** Python remains a dynamically typed language; type annotations are ignored at runtime but are checked by IDEs and linters (like `mypy`).

```python
def greet_user(name: str) -> str:
    return f"Hello, {name}!"

print(greet_user("Alice"))
```

### ⚡ Anonymous (Lambda) Functions
Lambda functions are small, one-line functions that do not have a name. They are defined using the `lambda` keyword.

#### 🔹 Syntax:
`lambda arguments: expression`

* They are limited to a single expression.
* They automatically return the result of that expression.
* Used for short-term, throwaway logic (e.g., in pandas `apply`, sorting, or higher-order functions).

```python
# Standard function
def square(x):
    return x * x

# Equivalent Lambda function
square_lambda = lambda x: x * x

print("Standard:", square(5))
print("Lambda:", square_lambda(5))
```

```python
# Lambda with multiple arguments
add_lambda = lambda a, b: a + b
print("Sum:", add_lambda(10, 20))
```

##### 🧠 Real-world Data Science / ML Usage: Sorting tuples/dictionaries by value

```python
# List of tuples containing (feature_name, importance_score)
feature_importance = [("Age", 0.15), ("Salary", 0.45), ("Location", 0.10), ("Education", 0.30)]

# Sort by importance score (second element in tuple)
feature_importance.sort(key=lambda item: item[1], reverse=True)
print("Sorted Feature Importance:", feature_importance)
```

### ⚙️ Higher-Order Functions (map, filter, reduce)
A higher-order function is a function that either takes a function as an argument, or returns a function.

#### 🔹 1. map()
* Applies a function to all items in an iterable.
* Syntax: `map(function, iterable)`

```python
numbers = [1, 2, 3, 4, 5]
squared = list(map(lambda x: x**2, numbers))
print("Squared Numbers:", squared)
```

#### 🔹 2. filter()
* Filters elements from an iterable based on whether a function returns `True` or `False`.
* Syntax: `filter(function, iterable)`

```python
scores = [85, 42, 90, 50, 78, 61]
passed = list(filter(lambda x: x >= 60, scores))
print("Passing Scores:", passed)
```

#### 🔹 3. reduce()
* Applies a function cumulatively to items in an iterable to reduce it to a single value.
* Must be imported from the `functools` module.

```python
from functools import reduce

nums = [1, 2, 3, 4, 5]
# Calculates: ((((1 * 2) * 3) * 4) * 5)
factorial_of_5 = reduce(lambda x, y: x * y, nums)
print("Factorial (using reduce):", factorial_of_5)
```

### 🔄 Recursion
Recursion is a programming technique where a function calls itself directly or indirectly to solve a problem.

#### 🔹 First Principles:
For a recursive function to work correctly and not crash the program, it must have two parts:
1. **Base Case:** The stopping condition that prevents infinite loops.
2. **Recursive Case:** The part where the function calls itself with a smaller input.

#### 🧠 The Call Stack & Stack Overflow
Each recursive call creates a new stack frame in memory. If the base case is missing or never reached, Python will raise a `RecursionError` (maximum recursion depth exceeded) to prevent stack overflow.

```python
# Factorial using Recursion: n! = n * (n-1)!
def factorial(n):
    # 1. Base Case
    if n == 0 or n == 1:
        return 1
    # 2. Recursive Case
    else:
        return n * factorial(n - 1)

print("Factorial of 5:", factorial(5))
```

```python
# Fibonacci Series using Recursion: F(n) = F(n-1) + F(n-2)
def fibonacci(n):
    # Base Cases
    if n <= 0:
        return 0
    elif n == 1:
        return 1
    # Recursive Case
    else:
        return fibonacci(n - 1) + fibonacci(n - 2)

print("6th Fibonacci number:", fibonacci(6))
```

##### 🧠 ML/AI Relevance:
* Tree-based algorithms (like Decision Trees, Random Forests, Gradient Boosting) are naturally built using recursive split processes on datasets.
* Parsing hierarchical data structures (like ASTs, JSON configuration trees, HTML parse trees) is most cleanly achieved using recursion.

### 🔗 Introduction to Nested Functions & Closures
A nested function is a function defined inside another function.

#### 🔹 What is a Closure?
A closure is a nested function that remembers and has access to variables in its enclosing (outer) scope, even after the outer function has finished executing.

#### 🔹 Requirements for a Closure:
1. We must have a nested function.
2. The nested function must refer to a value defined in the enclosing function.
3. The enclosing function must return the nested function.

```python
def power_factory(exponent):
    # Outer function scope variable: exponent
    def power_function(base):
        # Inner function has access to exponent
        return base ** exponent
    return power_function

# Create specific power functions
square_func = power_factory(2)
cube_func = power_factory(3)

print("5 squared:", square_func(5))
print("5 cubed:", cube_func(5))
```
