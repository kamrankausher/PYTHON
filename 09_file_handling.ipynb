### 🔹 1. File Opening Modes
Python provides several modes to open files depending on what operation you want to perform:

| Mode | Name | Description |
| :---: | :--- | :--- |
| **`'r'`** | Read | Default mode. Opens file for reading. Raises `FileNotFoundError` if file doesn't exist. |
| **`'w'`** | Write | Opens file for writing. Overwrites existing contents or creates a new file. |
| **`'a'`** | Append | Opens file for writing, appending to the end of the file. Creates file if missing. |
| **`'x'`** | Exclusive Create | Creates a new file. Raises `FileExistsError` if the file already exists. |
| **`'b'`** | Binary | Opens file in binary mode (e.g. `'rb'`, `'wb'`) for images, serialized models, etc. |
| **`'+'`** | Update | Opens file for both reading and writing (e.g., `'r+'`, `'w+'`). |

```python
try:
    # Exclusive creation mode
    file = open("exclusive_file.txt", "x")
    file.write("This file is exclusively created.")
    file.close()
    print("File created successfully.")
except FileExistsError:
    print("Error: The file already exists!")
```

***

### 🔹 2. Reading & Writing Methods
* **`read(n)`**: Reads the entire file or up to `n` characters/bytes.
* **`readline()`**: Reads a single line from the file.
* **`readlines()`**: Reads all lines into a list of strings.
* **`writelines(list)`**: Writes a list of strings to a file.

```python
# Creating a multi-line file
file = open("lines.txt", "w")
file.writelines(["Line 1\n", "Line 2\n", "Line 3\n"])
file.close()

# Reading line by line
file = open("lines.txt", "r")
print("First line:", file.readline().strip())
print("Remaining lines as list:", file.readlines())
file.close()
```

***

### 🔹 3. File Pointer Manipulation (seek & tell)
* **`tell()`**: Returns the current position of the file pointer (cursor) in terms of byte offset.
* **`seek(offset, whence)`**: Moves the file pointer to a new position.
  - `whence = 0` (default): Offset from the beginning of the file.
  - `whence = 1`: Offset from the current cursor position.
  - `whence = 2`: Offset from the end of the file.

```python
file = open("lines.txt", "r")
print("Initial cursor position:", file.tell()) # Should be 0

# Read some characters
data = file.read(5)
print("Read data:", data)
print("Cursor position after reading:", file.tell())

# Reset pointer to start
file.seek(0)
print("Cursor position after seek(0):", file.tell())
print("Reread start:", file.read(5))
file.close()
```

***

### 🔹 4. The with Statement (Context Managers)
* In real-world software, closing files manually is prone to bugs. If an exception occurs before `.close()`, the file descriptor remains open (causing memory leaks/file locks).
* The **`with`** statement defines a context manager that automatically handles opening and closing files safely, even if errors occur inside the block.

```python
with open("lines.txt", "r") as file:
    for line in file:
        print(line.strip())
# File is automatically closed here!
print("Is file closed?", file.closed)
```

### 🔹 5. Interacting with Structured Formats (JSON & CSV)
In AI & ML, data is rarely stored in plain text files. We frequently interact with JSON (configs/metadata) and CSV (tabular datasets).

#### 🔹 1. Working with JSON (`json` module)
* **`json.dump(obj, file)`**: Serializes Python object to file.
* **`json.load(file)`**: Deserializes JSON file to Python object.
* **`json.dumps(obj)`**: Serializes Python object to a string.
* **`json.loads(string)`**: Deserializes string to Python object.

```python
import json

config = {
    "model_name": "ResNet50",
    "learning_rate": 0.001,
    "batch_size": 32,
    "optimizer": "Adam"
}

# Write JSON
with open("config.json", "w") as f:
    json.dump(config, f, indent=4)

# Read JSON
with open("config.json", "r") as f:
    loaded_config = json.load(f)
print("Loaded Config Type:", type(loaded_config))
print("Learning Rate:", loaded_config["learning_rate"])
```

#### 🔹 2. Working with CSV (`csv` module)
* `csv.reader` / `csv.writer`: Read/write CSV files using lists.
* `csv.DictReader` / `csv.DictWriter`: Read/write CSV records as dictionaries (very intuitive).

```python
import csv

# Writing CSV using DictWriter
with open("dataset.csv", "w", newline="") as f:
    writer = csv.DictWriter(f, fieldnames=["id", "label", "confidence"])
    writer.writeheader()
    writer.writerow({"id": 1, "label": "Cat", "confidence": 0.98})
    writer.writerow({"id": 2, "label": "Dog", "confidence": 0.89})

# Reading CSV using DictReader
with open("dataset.csv", "r") as f:
    reader = csv.DictReader(f)
    for row in reader:
        print(f"ID: {row['id']} | Prediction: {row['label']} | Score: {row['confidence']}")
```

### 🔹 6. Modern Path Handling with `pathlib`
* Historically, Python developers used the `os.path` module for path manipulations (e.g. `os.path.join()`).
* Modern Python uses **`pathlib`**, which provides object-oriented, cross-platform path operations.

```python
from pathlib import Path

# Create a Path object representing current directory
current_dir = Path(".")
print("Absolute Path:", current_dir.resolve())

# Joining paths cleanly using the `/` operator
config_path = current_dir / "config.json"
print("Does config file exist?", config_path.exists())
print("File name:", config_path.name)
print("File suffix (extension):", config_path.suffix)
```
