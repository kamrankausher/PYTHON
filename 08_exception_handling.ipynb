## Exception Handling

````python
try:
    # Code that might cause an exception
    num1 = int(input("Enter numerator: "))
    num2 = int(input("Enter denominator: "))
    result = num1 / num2
    print(f"Result: {result}")
except Exception as e:
    # Executed if any exception occurs
    print(f"An error occurred: {e}")
````

***

### 🔹 Handling Specific Exceptions
* Using a bare `except:` or catching a general `Exception` is a bad practice. It can hide programming bugs and make debugging very difficult.
* It is always recommended to catch specific exceptions (like `ZeroDivisionError`, `ValueError`, `TypeError`, etc.).

````python
try:
    num1 = int(input("Enter numerator: "))
    num2 = int(input("Enter denominator: "))
    result = num1 / num2
    print(f"Result: {result}")
except ZeroDivisionError:
    print("Error: Denominator cannot be zero!")
except ValueError:
    print("Error: Please enter valid integers!")
except Exception as e:
    print(f"Unexpected error: {type(e).__name__} - {e}")
````

#### 🔹 Catching Multiple Exceptions in One Block
You can catch multiple exceptions together using a tuple.
```python
except (ValueError, ZeroDivisionError) as e:
    print(f"Invalid input or math error: {e}")
```

***

### 🔹 The else and finally Blocks
To build bulletproof code, Python provides two additional blocks: `else` and `finally`.

| Block | When does it run? | Main Purpose |
| :--- | :--- | :--- |
| **`else`** | Only if the `try` block succeeds (no exceptions thrown). | Place code that should run only if the try block succeeded. |
| **`finally`** | **Always** (whether exceptions occur or not, even after returns/breaks). | Resource cleanup (closing files, database links, releasing locks). |

````python
try:
    print("Opening database connection...")
    file = open("sample.txt", "w")
    data = 10 / int(input("Enter number: "))
except ZeroDivisionError:
    print("Math error occurred.")
except ValueError:
    print("Value error occurred.")
else:
    print("Calculated successfully! Data:", data)
    file.write(f"Result: {data}")
finally:
    print("Executing cleanup... Closing database / files.")
    file.close()
````

### 🔹 Raising Exceptions (raise) & Exception Chaining
Sometimes you want to intentionally trigger an exception when certain business rules or conditions are violated.

#### 🔹 1. Raising an Exception
Use the `raise` keyword followed by an exception class or instance.

````python
def set_age(age):
    if age < 0:
        raise ValueError("Age cannot be negative!")
    print(f"Age set to: {age}")

try:
    set_age(-5)
except ValueError as e:
    print("Caught expected exception:", e)
````

#### 🔹 2. Exception Chaining (`raise ... from`)
* When handling one exception, you might want to raise another exception while keeping track of the original cause.
* This is extremely useful in ML APIs to wrap low-level library errors into clear, domain-specific exceptions.

````python
def fetch_model_weights():
    try:
        # Simulating file reading exception
        open("weights.bin", "r")
    except FileNotFoundError as original_err:
        # Wrap into an custom/domain-specific error
        raise RuntimeError("Failed to load model weights.") from original_err

try:
    fetch_model_weights()
except RuntimeError as e:
    print("Chained Exception caught:", e)
    print("Original Cause:", e.__cause__)
````

### 🔹 Custom / User-Defined Exceptions
In large applications, built-in exceptions like `ValueError` or `RuntimeError` might be too generic. You can define custom exception classes by inheriting from the built-in `Exception` class.

#### 🧠 Best Practices:
* End the exception class name with `Error` (e.g. `InvalidSensorReadingError`).
* Keep inheritance simple (inherit from `Exception` or a base module exception).

````python
class DataValidationError(Exception):
    """Exception raised for errors in data preprocessing quality validation."""
    def __init__(self, column_name, message="Data validation failed"):
        self.column_name = column_name
        self.message = f"{message} in column '{column_name}'"
        super().__init__(self.message)

def check_null_values(df_col_name, has_nulls):
    if has_nulls:
        raise DataValidationError(df_col_name, "Missing values detected in critical feature")
    print("Data validated successfully!")

try:
    check_null_values("Age", has_nulls=True)
except DataValidationError as e:
    print(f"Pipeline Alert: {e}")
    print(f"Target column needing cleaning: {e.column_name}")
````
