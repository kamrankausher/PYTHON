### 🔹 1. The collections Module
Python's built-in dictionaries, lists, and tuples are highly versatile. However, the **`collections`** module provides specialized container data types designed for high-performance and clean coding.

#### 🔹 1. `Counter`
A subclass of `dict` designed to count hashable objects. It maps items to their counts dynamically.

```python
from collections import Counter

words = ["apple", "banana", "apple", "cherry", "banana", "apple"]
word_counts = Counter(words)

print("Word Counts Dictionary:", word_counts)
print("Count of 'apple':", word_counts["apple"])
# Fetch the top 2 most common items
print("Top 2 most common:", word_counts.most_common(2))
```

#### 🔹 2. `defaultdict`
* A standard dictionary throws a `KeyError` if you access a missing key.
* A `defaultdict` calls a factory function (like `int`, `list`, `set`) to provide a default value automatically when a missing key is accessed.
* This eliminates tedious check-and-initialize logic (`if key not in dict: dict[key] = ...`).

```python
from collections import defaultdict

# Grouping items by their category
grouped_data = defaultdict(list)

grouped_data["fruit"].append("apple")
grouped_data["fruit"].append("banana")
grouped_data["vegetable"].append("carrot")

print("Grouped Data:", dict(grouped_data))
print("Accessing empty key yields default list:", grouped_data["meat"])
```

#### 🔹 3. `deque` (Double-ended queue)
* Lists in Python are optimized for fast appends and pops at the **end** (O(1)). Appending or popping at the **beginning** of a list takes O(N) time because all other elements must be shifted in memory.
* `deque` provides thread-safe, memory-efficient appends and pops from both sides in O(1) time.

```python
from collections import deque

queue = deque(["user1", "user2"])

# Fast append to right and left
queue.append("user3")
queue.appendleft("admin")
print("Queue State:", queue)

# Fast pop from right and left
print("Popped from left:", queue.popleft())
print("Popped from right:", queue.pop())
print("Final Queue State:", queue)
```

#### 🔹 4. `namedtuple`
Provides a way to create lightweight, immutable database-like records (tuples) where fields can be accessed by index or by name (similar to properties).

```python
from collections import namedtuple

# Define a Point namedtuple
Point = namedtuple("Point", ["x", "y"])

p = Point(10, 20)
print(f"Point coordinates: x={p.x}, y={p.y}")
print(f"Index access: x={p[0]}, y={p[1]}")
```

### 🔹 2. The itertools Module
The **`itertools`** module contains a collection of fast, memory-efficient tools for working with iterators and looping.

#### 🔹 1. Infinite Iterators
* `count(start, step)`: Returns an infinite sequence of numbers.
* `cycle(iterable)`: Cycles through an iterable infinitely.
* `repeat(val, times)`: Repeats a value a set number of times.

#### 🔹 2. Terminating Iterators
* `chain(*iterables)`: Chains multiple iterables together sequentially.
* `accumulate(iterable)`: Returns running totals or accumulated results of a binary function.

```python
import itertools

# 1. chain() - Combine lists without copying elements
list1 = [1, 2]
list2 = [3, 4]
combined = list(itertools.chain(list1, list2))
print("Chained:", combined)

# 2. accumulate() - Running sums
nums = [1, 2, 3, 4]
print("Accumulated sums:", list(itertools.accumulate(nums)))
```

#### 🔹 3. Combinatoric Iterators
Used to compute permutations and combinations (frequently needed in probability, optimization, and search algorithms).
* `permutations(iterable, r)`: Unique ordered arrangements of length `r`.
* `combinations(iterable, r)`: Unique unordered selections of length `r`.

```python
items = ["A", "B", "C"]

# Combinations of 2 items
combos = list(itertools.combinations(items, 2))
print("Combinations (2):", combos)

# Permutations of 2 items
perms = list(itertools.permutations(items, 2))
print("Permutations (2):", perms)
```

#### 🔹 4. `groupby(iterable, key_func)`
Groups consecutive items in an iterable that share the same key.
* **Important Note:** The input iterable **must be sorted** by the grouping key first for `groupby` to work correctly.

```python
data = [
    {"name": "Alice", "dept": "HR"},
    {"name": "Bob", "dept": "IT"},
    {"name": "Charlie", "dept": "HR"},
    {"name": "David", "dept": "IT"}
]

# Sort by dept before grouping
data.sort(key=lambda x: x["dept"])

# Group by dept
for key, group in itertools.groupby(data, key=lambda x: x["dept"]):
    print(f"Department: {key}")
    for item in group:
        print(f"  - {item['name']}")
```

### 🔹 3. The functools Module
The **`functools`** module contains higher-order functions to manipulate functions themselves.

#### 🔹 1. `partial(func, *args, **kwargs)`
Allows you to "freeze" a portion of a function's arguments, creating a new function with a simpler signature.

```python
from functools import partial

# A generic multiplier function
def multiply(x, y):
    return x * y

# Create a specialized 'double' function by freezing x=2
double = partial(multiply, 2)
triple = partial(multiply, 3)

print("Double of 5:", double(5))
print("Triple of 5:", triple(5))
```

#### 🔹 2. `lru_cache`
As seen in Chapter 12, `@lru_cache` memoizes function call results. We can clear its memory using `function_name.cache_clear()` when data changes.

```python
from functools import lru_cache

@lru_cache(maxsize=3)
def fetch_user_settings(user_id):
    print(f"[DB] Fetching settings for user {user_id}...")
    return {"id": user_id, "theme": "dark"}

# First calls -> triggers DB print
fetch_user_settings(1)
fetch_user_settings(2)

# Second calls -> returned from memory cache
fetch_user_settings(1)

# Clear cache memory
fetch_user_settings.cache_clear()
print("Cache cleared.")

# Will hit database again
fetch_user_settings(1)
```
