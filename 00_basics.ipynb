#### How Python Executes Code?
-> You write .py file

-> Python converts code → bytecode

-> Bytecode runs on Python Virtual Machine (PVM)

#### Interpreted vs Compiled: 
##### Compiled language (C++)

1. Code → Machine code → Run
2. Faster execution
3. Harder to debug

##### Interpreted language (Python)

1. Code executed line by line
2. Easier debugging
3. Slower speed

**#Important note :** Python is:

##### Both compiled (to bytecode) AND interpreted (by PVM)

### 🔴 Hard Level


```python
#Q23 Write a program to redirect standard output (stdout) to a file named 'output.txt' temporarily and print 'Hello File'.
import sys
original_stdout = sys.stdout
with open('output.txt', 'w') as f:
    sys.stdout = f
    print('Hello File')
sys.stdout = original_stdout
print('Redirected back to console. Check output.txt!')
```

```python
#Q24 Write a program to measure the exact execution time of a print statement in microseconds.
import time
start = time.perf_counter_ns()
print('Benchmarked statement')
end = time.perf_counter_ns()
print(f'Execution Time: {(end - start)/1000:.3f} microseconds')
```

```python
#Q26 Write a program to count total memory usage of a string as its length increases from 1 to 100 characters.
import sys
for length in [1, 10, 50, 100]:
    s = 'a' * length
    print(f'Length: {length} -> Memory Size: {sys.getsizeof(s)} bytes')
```

```python
#Q27 Write a program to print all the system paths in sys.path line-by-line.
import sys
for path in sys.path[:5]:  # Print first 5 for preview
    print(path)
```

```python
#Q28 Write a program to implement a custom printing function using sys.stdout.write instead of print.
import sys
def custom_print(text):
    sys.stdout.write(str(text) + '\n')
    sys.stdout.flush()
custom_print('Hello from sys.stdout.write!')
```

```python
#Q29 Write a program to calculate compound interest using user inputs for Principal, Rate, Time, and frequency.
p = float(input('Principal: '))
r = float(input('Rate (%): '))
t = float(input('Time (years): '))
n = int(input('Compounding frequency: '))
a = p * (1 + (r / 100) / n) ** (n * t)
print(f'Compound Interest: {a - p:.2f}, Total Amount: {a:.2f}')
```
