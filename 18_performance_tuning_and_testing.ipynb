### 🔹 1. Unit Testing with pytest
Writing automated tests ensures that your functions work correctly as you add new features, preventing regressions. While Python has a built-in `unittest` module, **`pytest`** is the industry standard due to its simple syntax and powerful features.

#### 🔹 Key Features of pytest:
* Simple assert statements: No need for `self.assertEqual()`, just use standard Python `assert`.
* **Fixtures:** Reusable setup operations (like creating a database connection or loading a dataset) that can be passed as function arguments.
* **Parameterization:** Run a test function with multiple sets of test inputs dynamically.

#### How to write a pytest test file (e.g. `test_math.py`):
```python
def add(x, y):
    return x + y

def test_add():
    assert add(2, 3) == 5
    assert add(-1, 1) == 0
```
Run tests in your terminal using:
```bash
pytest test_math.py
```

````python
# Writing a pytest test fixture simulation
import pytest

# Mocking a dataset loader fixture
@pytest.fixture
def sample_dataset():
    return [
        {"id": 1, "label": "Cat", "score": 0.95},
        {"id": 2, "label": "Dog", "score": 0.85}
    ]

def filter_high_confidence(dataset, threshold=0.9):
    return [row for row in dataset if row["score"] >= threshold]

def test_filter(sample_dataset):
    filtered = filter_high_confidence(sample_dataset, 0.9)
    assert len(filtered) == 1
    assert filtered[0]["label"] == "Cat"

# Run it inside notebook (requires ipytest or similar, here we just show the structure)
print("Fixture and test structures defined successfully.")
````

### 🔹 2. Mocking External Dependencies (unittest.mock)
In unit testing, you should isolate the code under test. If a function makes an external HTTP request, queries a live database, or reads a huge file, you should **mock** (simulate) that dependency to keep tests fast, reliable, and offline.

* Python provides `unittest.mock` with the **`Mock`** class and the **`patch`** decorator.

````python
from unittest.mock import Mock, patch

def get_exchange_rate(currency):
    # Simulating a live API call that could fail or be slow
    import urllib.request
    import json
    url = f"https://api.exchangerate.host/latest?base={currency}"
    response = urllib.request.urlopen(url)
    return json.loads(response.read())["rates"]["USD"]

# Mocking the api call
@patch("urllib.request.urlopen")
def test_get_exchange_rate(mock_urlopen):
    # Create a mock response object
    mock_response = Mock()
    mock_response.read.return_type = b'{"rates": {"USD": 1.25}}'
    mock_urlopen.return_type = mock_response
    
    # We call the function and verify it returns mocked rates
    # rate = get_exchange_rate("GBP")
    # assert rate == 1.25
    print("Mock urlopen set up correctly in test suite.")

test_get_exchange_rate()
````

### 🔹 3. Performance Profiling (timeit & cProfile)
Before optimizing code, you must measure its execution time to find the actual bottlenecks. Optimizing without profiling is called **premature optimization** and is an anti-pattern.

#### 🔹 1. `timeit` (Micro-benchmarking)
Used to measure the execution time of small code snippets. Runs the code thousands of times to get an accurate average.

````python
import timeit

# Compare list comprehension vs loop append
comp_time = timeit.timeit("[x**2 for x in range(1000)]", number=1000)
loop_time = timeit.timeit("""
result = []
for x in range(1000):
    result.append(x**2)
""", number=1000)

print(f"List Comprehension: {comp_time:.5f} seconds")
print(f"For Loop Append:    {loop_time:.5f} seconds")
````

#### 🔹 2. `cProfile` (Macro-benchmarking)
A built-in performance profiler that counts function calls and measures execution times of different functions within a program.

````python
import cProfile

def heavy_math():
    result = []
    for i in range(1000000):
        result.append(i * i)
    return result

def main_app():
    print("Starting process...")
    data = heavy_math()
    print("Finished.")

# Profile the execution
cProfile.run("main_app()")
````

### 🔹 4. Basic Speed Optimization Strategies
When writing performance-sensitive Python code, keep these rules of thumb in mind:

1. **Avoid Globals:** Accessing local variables is faster than global variables in Python because local variables are stored in an array while globals require a dictionary lookup.
2. **Leverage Built-ins:** Python's built-in functions (like `sum()`, `map()`, `any()`, `all()`, and methods in `itertools`) are implemented in optimized C.
3. **List Comprehensions over Loops:** List comprehensions run at C-speed inside the interpreter.
4. **Use Vectorization (NumPy):** For numeric math, avoid loops entirely. Use NumPy arrays which execute element-wise operations in compiled C/Fortran blocks.
5. **Numba JIT compiler:** For heavy loops, use `@jit` decorator from `numba` to compile Python code to machine instructions at runtime.

````python
# 1. Local variable access vs Global access comparison
import time

global_var = 10

def use_global():
    total = 0
    for _ in range(10000000):
        total += global_var # dict lookup
    return total

def use_local():
    local_var = 10
    total = 0
    for _ in range(10000000):
        total += local_var # array lookup
    return total

start = time.perf_counter()
use_global()
mid = time.perf_counter()
use_local()
end = time.perf_counter()

print(f"Global lookup time: {mid - start:.5f} seconds")
print(f"Local lookup time:  {end - mid:.5f} seconds")
````
